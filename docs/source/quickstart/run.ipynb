# Typical run

The typical run involves the full pipeline (`topic3d runall`), which includes loading and pre-processing the data, training an LDA model, and decomposing the data, all with a single command that passes both data files and parameters as arguments.

```python
topic3d runall \
    --data_train_file ./data/sample_sim_train.npy \
    --data_test_file ./data/sample_sim_test.npy \
    --results_dir ./results/ \
    --num_components 5 \
    --max_iter 100 \
    --random_state 64
```

For information on the parameters and their accepted values or formats, see Reference > Parameters.

A detailed workflow is available in User Guide > Workflow
