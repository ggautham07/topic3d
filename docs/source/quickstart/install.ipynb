# Installing topic3d

This notebook provides a quick guide to installing **topic3d** and its dependencies.

## Overview

`topic3d` is a Python package designed for topic modelling of single-cell chromatin tracing proximity datasets.

This guide covers:
- Setting up a virtual environment
- Installation via pip
- Verifying the installation

## Requirements

- Python 3.9–3.13 recommended
- conda and pip

## 1. Create a virtual environment

Install [conda](https://docs.conda.io/projects/conda/en/stable/index.html) or the lighter version [miniconda](https://www.anaconda.com/docs/getting-started/miniconda/main) and create a virtual environment using this [file](https://github.com/ggautham07/topic3d/blob/dev/environment-dev.yml). This YML file was exported from a stable, well-tested environment.

```python
conda create -n topic3d-env --file "<path_to_file>"
```

The environment can be activated using

```python
conda activate topic3d-env
```

## 2. Install topic3d

We will use pip to install topic3d within this environment. Since all dependecies will have already been installed in the previous step, this should be quick.

```python
pip install topic3d
```

### Optional: Install from source

```python
git clone https://github.com/ggautham07/topic3d.git
cd topic3d
pip install .
```

## 4. Check CLI

```python
!topic3d -h
```

## Troubleshooting

- Use a clean environment
- Check Python version
- Reinstall if needed
