# Installing topic3d

This notebook provides a quick guide to installing **topic3d** and its dependencies.

## Overview

`topic3d` is a Python package designed for topic modelling of single-cell chromatin tracing proximity datasets.

This guide covers:
- Installation via pip
- Setting up a virtual environment
- Verifying the installation

## Requirements

- Python 3.9–3.13 recommended
- pip or conda

## 1. Create a virtual environment (recommended)

```python
conda create -n topic3d-env python=3.11
conda activate topic3d-env
```

Or using venv:

```python
python -m venv topic3d-env
source topic3d-env/bin/activate  # Linux/macOS
topic3d-env\Scripts\activate   # Windows
```

## 2. Install topic3d

```python
pip install topic3d
```

### Optional: Install from source

```python
git clone https://github.com/ggautham07/topic3d.git
cd topic3d
pip install -e .
```

## 3. Verify installation

```python
import topic3d
print(topic3d.__version__)
```

## 4. Check CLI

```python
topic3d runall --help
```

## Troubleshooting

- Use a clean environment
- Check Python version
- Reinstall if needed
